```python
import pandas as pd
import numpy as np
```

```python
from arff_csv import arff_to_csv
df = arff_to_csv("data_plant.arff", "data.csv")
df = pd.DataFrame(df)
df.head()
```

```python
df.info()
```

```python
x = df.drop(columns = "Class")
y = df["Class"]
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
from sklearn.model_selection import train_test_split
xtrain , xtest , ytrain , ytest = train_test_split(x,y,test_size=0.2,random_state= 85)
```

```python
from lazypredict.Supervised import LazyClassifier
clf = LazyClassifier()
models ,gout= clf.fit(xtrain,xtest,ytrain,ytest)
models
```

```python
hot = list(df.columns)
hot = hot[:-1:]
print(hot)
```

```python
from sklearn.ensemble import ExtraTreesClassifier
from sklearn.pipeline import make_pipeline
from sklearn.compose import make_column_transformer

cl = make_column_transformer((StandardScaler(),hot),remainder="passthrough")
pipi = make_pipeline(cl,ExtraTreesClassifier())
```

```python
pipi.fit(xtrain,ytrain)
```

```python
yhat = pipi.predict(xtest)
from sklearn.metrics import classification_report as cr 
print(cr(ytest,yhat))
```

```python
print(cr(ytrain,pipi.predisct(xtrain)))
```
