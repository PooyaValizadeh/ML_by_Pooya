```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def sigmoid(x):
    return 1 / (1+np.exp(-x))
def sigmoid_moshtag(x):
    s = sigmoid(x)
    return s * (1 - s)
```

```python
def relu(x):
    return np.maximum(0,x)
def relu_moshtag(x):
    return np.where(x > 0, 1.0, 0.0)
```

```python
def elu(x,alpha=1.0):
     return np.where(x > 0, x, alpha * (np.exp(x) - 1))
def elu_moshtag(x, alpha=1.0):
    return np.where(x > 0, 1.0, alpha * np.exp(x))
```

```python
SCALE = 1.0507009873554804934193349852946
ALPHA = 1.6732632423543772848170429916717

def selu(x, scale=SCALE, alpha=ALPHA):
    return scale * np.where(x > 0, x, alpha * (np.exp(x) - 1))

def selu_moshtag(x, scale=SCALE, alpha=ALPHA):
    return scale * np.where(x > 0, 1.0, alpha * np.exp(x))
```

```python
def tanh(x):
    return np.tanh(x)
def tanh_moshtag(x):
    return 1- np.tanh(x) ** 2
```

```python
x = np.linspace(-10, 10, 1000)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].plot(x, sigmoid(x),      label='Sigmoid')
axes[0].plot(x, tanh(x),         label='Tanh')
axes[0].plot(x, relu(x),         label='ReLU')
axes[0].plot(x, elu(x),          label='ELU')
axes[0].plot(x, selu(x),         label='SELU')

axes[0].axhline(0, color='gray', linewidth=0.5, linestyle='--')
axes[0].axvline(0, color='gray', linewidth=0.5, linestyle='--')
axes[0].set_xlabel('x')
axes[0].set_ylabel('f(x)')
axes[0].legend()
axes[0].grid(True, alpha=0.3)
axes[1].plot(x, sigmoid_moshtag(x), label="Sigmoid'")
axes[1].plot(x, tanh_moshtag(x),    label="Tanh'")
axes[1].plot(x, relu_moshtag(x),    label="ReLU'")
axes[1].plot(x, elu_moshtag(x),     label="ELU'")
axes[1].plot(x, selu_moshtag(x),    label="SELU'")
axes[1].axhline(0, color='gray', linewidth=0.5, linestyle='--')
axes[1].axvline(0, color='gray', linewidth=0.5, linestyle='--')
axes[1].set_xlabel('x')
axes[1].set_ylabel("f'(x)")
axes[1].legend()
axes[1].grid(True, alpha=0.3)
plt.show()
```
