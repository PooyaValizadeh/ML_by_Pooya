```python
import tensorflow as tf
from tensorflow import keras
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
(xtrain , ytrain) , (xtest , ytest) = keras.datasets.cifar10.load_data()
```

```python
images = xtrain[:10] / 255
```

```python
images.shape
```

```python
filters= np.zeros(shape=(2,3,3,3),dtype = np.float32)
```

```python
filters[0,:,0,0]  = 1
filters[0,:,1,1] = 1
filters[0,:,2,2] = 1

filters[1,:,0,0]  = 1
filters[1,:,1,1] = 1
filters[1,:,2,2] = 1

# filters[1,:,0,2]  = 1
# filters[1,:,1,1] = 1
# filters[1,:,2,0] = 1

filters
```

```python
output = tf.nn.conv2d(images,filters,strides = 1 , padding="SAME")
```

```python
output.shape
```

```python
for i in range(10):
    plt.imshow(output[i,:,:,:])
    plt.show()
```
