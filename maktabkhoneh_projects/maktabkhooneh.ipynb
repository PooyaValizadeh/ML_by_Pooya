# پروژه پویا ولی زاده
## با عرض سلام و ادب خدمت منتور محترم. 
من کمی متفاوت این پروژه رو انجام دادم چون دیگر نمی شود از خلاصه گر ترنسفرمر استفاده کرد و باید مستقیم یک مدل خلاصه گر را ایمپورت کرد پس من هم همین کار را انجام دادم. همچنین مایل بودم پروژه رو روی کگل اپلود کنم ولی برای ایران تحریم است

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import requests
from bs4 import BeautifulSoup
```

```python
URL = "https://medium.com/analytics-vidhya/openai-gpt-3-language-models-are-few-shot-learners-82531b3d3122"
r = requests.get(URL)
```

```python
soup = BeautifulSoup(r.text, 'html.parser')
results = soup.find_all(['h1', 'p'])
text = [result.text for result in results]
ARTICLE = ' '.join(text)
```

```python
ARTICLE
```

```python
max_chunk = 500
ARTICLE = ARTICLE.replace('.', '')
ARTICLE = ARTICLE.replace('?', '')
ARTICLE = ARTICLE.replace('!', '')
sentences = ARTICLE.split(' ')
current_chunk = 0 
chunks = []
for sentence in sentences:
    if len(chunks) == current_chunk + 1: 
        if len(chunks[current_chunk]) + len(sentence.split(' ')) <= max_chunk:
            chunks[current_chunk].extend(sentence.split(' '))
        else:
            current_chunk += 1
            chunks.append(sentence.split(' '))
    else:
        print(current_chunk)
        chunks.append(sentence.split(' '))

for chunk_id in range(len(chunks)):
    chunks[chunk_id] = ' '.join(chunks[chunk_id])
```

```python

model = "Falconsai/text_summarization"
tokenizer = AutoTokenizer.from_pretrained(model)
model1 = AutoModelForSeq2SeqLM.from_pretrained(model)


inputs = tokenizer(chunks, return_tensors="pt", truncation=True, max_length=512)


summary1 = model1.generate(
    inputs["input_ids"],
    max_length=120,
    min_length=50,
    do_sample=False,
)

```

## خروجی
