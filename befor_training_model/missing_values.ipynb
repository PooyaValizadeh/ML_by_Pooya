```python
import pandas as pd
import numpy as np
from sklearn.compose import make_column_transformer
```

```python
df = pd.DataFrame({"data1" : [12,34,2,5,np.nan],
"data2" : [np.nan,15,34,23,1]})
df
```

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import SimpleImputer , IterativeImputer , KNNImputer
cl = make_column_transformer((SimpleImputer(),["data1"]),remainder='passthrough')
df1 = cl.fit_transform(df)
df1
```

```python
it = IterativeImputer()
df2 = it.fit_transform(df)
df2
```

```python
df3 = KNNImputer(n_neighbors=2).fit_transform(df)
df3
```
